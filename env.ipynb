# Simulation
Generates sensor stream to train the AI model


1. RainfallEngine    -> stochastic, seasonally-correct storm events
                          (uses MONTHLY_DATA, EVENT_*, STORM_* tables)
2. EnvironmentEngine  -> temperature / humidity / weather state / E
                          (uses MONTHLY_CLIMATE, WEATHER_*, CLIMATE_REFERENCE)
3. DemandEngine       -> analytic human water-demand curve Q(t,E)
                          (uses DEMAND_MODEL_PARAMS, NORMALIZATION_CONSTANT)
4. DrainSimulator     -> converts rainfall-runoff + sewage demand into
                          a single hydraulic inflow -> smoothed flow ->
                          radar water-level reading (cm), with sensor noise.

```python
from __future__ import annotations
import calendar
import math
import random
import time
from dataclasses import dataclass
from typing import Dict, List, Optional, Tuple
import numpy as np
import pandas as pd
# import matplotlib.pyplot as plt

from Factors import *
```

```python
SIM_CONFIG = {
    "timestep_minutes": 5,     # internal physics resolution
    "sensor_interval_minutes": 15,  # how often the radar sensor actually reports
    "extra_event_prob": 0.15,  # chance a rain-day gets a 2nd event
    "max_events_per_day": 2,
}


@dataclass
class DrainConfig:
    """Physical / hydraulic configuration of the single monitored drain."""
    name: str = "Manhole_1"
    catchment_area_m2: float = 12000.0     # tributary paved area feeding this drain
    runoff_coefficient: float = 0.75       # fraction of rain that becomes runoff
    base_sewage_lps: float = 6.0           # avg domestic flow (L/s) at demand_index=1
    pipe_capacity_lps: float = 350.0       # flow at which the level curve saturates
    max_level_cm: float = 100.0            # manhole depth visible to the radar sensor
    level_time_constant_min: float = 6.0   # hydraulic lag / storage smoothing
    sensor_noise_std_cm: float = 0.15      # radar sensor measurement noise (cm)
    flow_noise_frac: float = 0.02          # relative noise on sewage baseflow
```

### RainFall

```python
_MONTH_NAMES = list(calendar.month_name)[1:]

def _days_in_month_name(month: str, year: int = 2023) -> int:
    idx = _MONTH_NAMES.index(month) + 1
    return calendar.monthrange(year, idx)[1]


def classify_event_size(total_mm: float) -> str:
    """Bucket a storm total (mm) into one of EVENT_SIZE_CLASSES."""
    if total_mm < EVENT_SIZE_CLASSES["light"]["range_mm"][0]:
        return "light"
    for cls, spec in EVENT_SIZE_CLASSES.items():
        lo, hi = spec["range_mm"]
        if hi is None:
            if total_mm >= lo:
                return cls
        elif lo <= total_mm < hi:
            return cls
    return "extreme"


def generate_single_event(month: str, rng: random.Random, apply_calibration: bool = True) -> dict:
    """
    Build one rain event using:
      - EVENT_DURATION_DISTRIBUTION / DURATION_BUCKET_HOURS -> duration
      - intensity_range/typical + EVENT_INTENSITY_DISTRIBUTION_WITHIN_CLASS -> avg intensity
      - STORM_PROFILES / STORM_PROFILE_WEIGHTS / PEAK_MULTIPLIER_BY_EVENT_CLASS -> hyetograph shape
    Returns per-segment rainfall (mm) that sums exactly to the event total.
    """
    season = MONTH_TO_SEASON[month]
    mdata = MONTHLY_DATA[month]

    # --- duration ---
    dist = EVENT_DURATION_DISTRIBUTION[season]
    buckets, weights = zip(*dist.items())
    bucket = rng.choices(buckets, weights=weights, k=1)[0]
    lo_d, hi_d = DURATION_BUCKET_HOURS[bucket]
    duration_hr = rng.uniform(lo_d, hi_d)

    # --- avg intensity regime ---
    i_lo, i_hi = mdata["intensity_range_mmph"]
    i_typ = mdata["intensity_typical_mmph"]
    roll = rng.random()
    idist = EVENT_INTENSITY_DISTRIBUTION_WITHIN_CLASS
    if roll < idist["lower_half_pct"]:
        avg_intensity = rng.uniform(i_lo, i_typ)
    elif roll < idist["lower_half_pct"] + idist["upper_half_pct"]:
        avg_intensity = rng.uniform(i_typ, i_hi)
    else:
        avg_intensity = rng.uniform(i_hi, i_hi * 1.6)  # unusually intense outlier burst

    total_mm = avg_intensity * duration_hr
    event_class = classify_event_size(total_mm)

    # --- storm shape (hyetograph) ---
    profile_name = rng.choices(
        list(STORM_PROFILE_WEIGHTS.keys()),
        weights=list(STORM_PROFILE_WEIGHTS.values()), k=1,
    )[0]
    base_profile = np.array(STORM_PROFILES[profile_name], dtype=float)

    peak_lo, peak_hi = PEAK_MULTIPLIER_BY_EVENT_CLASS[event_class]
    target_peak = rng.uniform(peak_lo, peak_hi)
    scaled = base_profile * (target_peak / base_profile.max())
    shape = scaled / scaled.mean()  # renormalise so mean == 1 (preserves avg intensity)

    n_seg = len(shape)
    seg_duration_hr = duration_hr / n_seg
    seg_mm = shape * avg_intensity * seg_duration_hr  # sums exactly to total_mm

    if apply_calibration:
        factor = RAIN_CALIBRATION_FACTORS.get(month, 1.0)
        seg_mm = seg_mm * factor

    return {
        "duration_hr": duration_hr,
        "avg_intensity_mmph": avg_intensity,
        "event_class": event_class,
        "profile_name": profile_name,
        "seg_duration_hr": seg_duration_hr,
        "seg_mm": seg_mm,
        "total_mm": float(seg_mm.sum()),
    }


def generate_rain_day_events(month: str, rng: random.Random) -> List[dict]:
    """Decide (Bernoulli, rain_prob_pct) whether it rains today, and build events."""
    prob = MONTHLY_DATA[month]["rain_prob_pct"] / 100.0
    if rng.random() >= prob:
        return []

    events = [generate_single_event(month, rng)]
    if SIM_CONFIG["max_events_per_day"] > 1 and rng.random() < SIM_CONFIG["extra_event_prob"]:
        events.append(generate_single_event(month, rng))

    for ev in events:
        ev["start_min_of_day"] = rng.uniform(0, 24 * 60 - 1)
    return events


def _expected_mm_per_event(month: str, n_samples: int = 800, seed: int = 12345) -> float:
    rng = random.Random(seed)
    total = sum(
        generate_single_event(month, rng, apply_calibration=False)["total_mm"]
        for _ in range(n_samples)
    )
    return total / n_samples


def _build_calibration_factors() -> Dict[str, float]:
    """
    The granular tables give SHAPE (duration/intensity distributions) but not
    the exact per-event size-class weighting needed to hit total_rainfall_mm
    exactly. Computes a per-month multiplicative correction so the long-run
    simulated monthly total converges to MONTHLY_DATA[month]['total_rainfall_mm'].
    """
    factors = {}
    for month, mdata in MONTHLY_DATA.items():
        days = _days_in_month_name(month)
        p = mdata["rain_prob_pct"] / 100.0
        expected_events_per_day = 1.0 + SIM_CONFIG["extra_event_prob"]
        mm_per_event = _expected_mm_per_event(month)
        expected_total = days * p * expected_events_per_day * mm_per_event
        target_total = mdata["total_rainfall_mm"]
        factors[month] = (target_total / expected_total) if expected_total > 1e-6 else 1.0
    return factors


RAIN_CALIBRATION_FACTORS = _build_calibration_factors()


def _accumulate(grid: np.ndarray, seg_start_min: float, seg_end_min: float,
                seg_mm: float, step_min: float) -> None:
    """Rasterise a (possibly sub-timestep) rainfall segment onto the output grid."""
    if seg_mm <= 0:
        return
    seg_len = seg_end_min - seg_start_min
    if seg_len <= 0:
        return
    n = len(grid)
    i0 = int(seg_start_min // step_min)
    i1 = int(math.ceil(seg_end_min / step_min)) - 1
    if i1 < 0 or i0 >= n:
        return
    i0, i1 = max(i0, 0), min(i1, n - 1)
    for i in range(i0, i1 + 1):
        cell_start, cell_end = i * step_min, (i + 1) * step_min
        overlap = min(seg_end_min, cell_end) - max(seg_start_min, cell_start)
        if overlap > 0:
            grid[i] += seg_mm * (overlap / seg_len)


class RainfallEngine:
    def __init__(self, timestep_minutes: int):
        self.step = timestep_minutes

    def simulate(self, start_date, n_days: int, rng: random.Random) -> Tuple[np.ndarray, List[dict]]:
        n_steps = int(round(n_days * 24 * 60 / self.step))
        rain_mm = np.zeros(n_steps, dtype=float)
        event_log: List[dict] = []
        start_ts = pd.Timestamp(start_date)

        for day in range(n_days):
            date = start_ts + pd.Timedelta(days=day)
            month = date.month_name()
            day_start_min = day * 24 * 60

            for ev in generate_rain_day_events(month, rng):
                ev_start_min = day_start_min + ev["start_min_of_day"]
                seg_dur_min = ev["seg_duration_hr"] * 60
                for i, mm in enumerate(ev["seg_mm"]):
                    seg_start = ev_start_min + i * seg_dur_min
                    seg_end = seg_start + seg_dur_min
                    _accumulate(rain_mm, seg_start, seg_end, mm, self.step)

                event_log.append({
                    "date": date.date(), "month": month, "season": MONTH_TO_SEASON[month],
                    "start_min_of_sim": ev_start_min, "duration_hr": ev["duration_hr"],
                    "avg_intensity_mmph": ev["avg_intensity_mmph"],
                    "total_mm": ev["total_mm"], "event_class": ev["event_class"],
                    "profile": ev["profile_name"],
                })

        return rain_mm, event_log
```

### Environment

```python
class EnvironmentEngine:
    def __init__(self, np_rng: np.random.Generator):
        self.np_rng = np_rng

    def simulate(self, timeline: pd.DatetimeIndex, rain_intensity_mmph: np.ndarray) -> pd.DataFrame:
        n = len(timeline)
        hour = timeline.hour.values + timeline.minute.values / 60.0
        months = timeline.month_name()
        seasons = months.map(MONTH_TO_SEASON)

        temp_mean = months.map(lambda m: MONTHLY_CLIMATE[m]["temp_c_mean"]).values.astype(float)
        temp_range = months.map(lambda m: MONTHLY_CLIMATE[m]["temp_c_range"]).values.astype(float)
        rh_mean = months.map(lambda m: MONTHLY_CLIMATE[m]["rh_pct_mean"]).values.astype(float)
        rh_range = months.map(lambda m: MONTHLY_CLIMATE[m]["rh_pct_range"]).values.astype(float)

        # diurnal cycle: temp peaks ~15:00, RH troughs ~15:00
        phase = 2 * np.pi * (hour - 15.0) / 24.0
        temp_c = temp_mean + temp_range * np.cos(phase) + self.np_rng.normal(0, 0.3, n)
        rh_base = rh_mean - rh_range * np.cos(phase)

        weather_state = self._weather_states(timeline, rain_intensity_mmph)
        rh_adj = np.array([WEATHER_RH_ADJUSTMENT[w] for w in weather_state])
        rh_pct = np.clip(rh_base + rh_adj, *CLIMATE_REFERENCE["rh_pct_clip"])

        ref = CLIMATE_REFERENCE
        F_temperature = np.clip(
            1.0 + (temp_c - ref["temp_c_ref"]) / ref["temp_c_span"], *ref["F_temp_clamp"])
        F_humidity = np.clip(
            1.0 - (rh_pct - ref["rh_pct_ref"]) / ref["rh_pct_span"] * ref["rh_pct_effect"],
            *ref["F_rh_clamp"])
        F_season = seasons.map(F_SEASON_BY_KEY).values.astype(float)
        F_weather = np.array([WEATHER_STATES[w]["F_weather"] for w in weather_state])

        E = np.clip(F_temperature * F_humidity * F_season * F_weather, *ref["E_clamp"])

        return pd.DataFrame({
            "season": seasons.values,
            "temp_c": temp_c, "rh_pct": rh_pct, "weather_state": weather_state,
            "E": E,
        }, index=timeline)

    def _weather_states(self, timeline: pd.DatetimeIndex, rain_intensity: np.ndarray) -> np.ndarray:
        n = len(timeline)
        state = np.empty(n, dtype=object)
        rainy = rain_intensity > 1e-4

        conditions = [
            rain_intensity <= 1.0,
            rain_intensity <= 4.0,
            rain_intensity <= 10.0,
            rain_intensity > 10.0,
        ]
        choices = ["drizzle", "light_rain", "moderate_rain", "heavy_rain"]
        rainy_state = np.select(conditions, choices, default="light_rain")
        state[rainy] = rainy_state[rainy]

        # dry-day weather is drawn once per calendar day for temporal continuity
        day_ordinal = timeline.floor("D")
        unique_days = day_ordinal.unique()
        dry_keys = list(DRY_WEATHER_MIX.keys())
        dry_probs = list(DRY_WEATHER_MIX.values())
        dry_choice_by_day = {d: self.np_rng.choice(dry_keys, p=dry_probs) for d in unique_days}
        dry_state = day_ordinal.map(dry_choice_by_day).values
        state[~rainy] = dry_state[~rainy]
        return state
```

### Human Demand

```python
class DemandEngine:
    """Analytic Q(t, E) model — see DEMAND_MODEL_PARAMS / NORMALIZATION_CONSTANT."""

    @staticmethod
    def q_base(hour_of_day: np.ndarray) -> np.ndarray:
        p = DEMAND_MODEL_PARAMS
        morning = p["A1"] * np.exp(-((hour_of_day - p["mu1"]) ** 2) / (2 * p["sigma1"] ** 2))
        evening = p["A2"] * np.exp(-((hour_of_day - p["mu2"]) ** 2) / (2 * p["sigma2"] ** 2))
        raw = p["B"] + morning + evening
        return raw / NORMALIZATION_CONSTANT

    @classmethod
    def demand_index(cls, hour_of_day: np.ndarray, E: np.ndarray) -> np.ndarray:
        return E * cls.q_base(hour_of_day)
```

### Drain Water Level Simulation

```python
class DrainSimulator:
    """
    Converts rainfall intensity + human demand into a synthetic radar
    water-level reading for ONE drain. No upstream routing, no network.
    """

    def __init__(self, drain: DrainConfig, timestep_minutes: int, rng_seed: Optional[int] = None):
        self.drain = drain
        self.timestep_minutes = timestep_minutes
        self.np_rng = np.random.default_rng(rng_seed)

    def simulate(self, timeline: pd.DatetimeIndex, rain_intensity_mmph: np.ndarray,
                 demand_index: np.ndarray) -> pd.DataFrame:
        n = len(timeline)
        d = self.drain

        runoff_lps = rain_intensity_mmph * d.catchment_area_m2 * d.runoff_coefficient / 3600.0
        sewage_lps = d.base_sewage_lps * demand_index
        noise = self.np_rng.normal(0, d.flow_noise_frac * d.base_sewage_lps, n)
        inflow_lps = np.clip(runoff_lps + sewage_lps + noise, 0, None)

        # linear-reservoir smoothing (hydraulic lag / storage effect)
        inflow_series = pd.Series(inflow_lps, index=timeline)
        smoothed = inflow_series.ewm(
            halflife=pd.Timedelta(minutes=d.level_time_constant_min), times=timeline
        ).mean()

        # saturating flow -> level relationship (pipe fills up asymptotically)
        level_cm = d.max_level_cm * (1.0 - np.exp(-smoothed.values / d.pipe_capacity_lps))
        sensor_noise = self.np_rng.normal(0, d.sensor_noise_std_cm, n)
        level_cm = np.clip(level_cm + sensor_noise, 0, d.max_level_cm)

        return pd.DataFrame({
            "inflow_lps": inflow_lps,
            "smoothed_flow_lps": smoothed.values,
            "level_cm": level_cm,
        }, index=timeline)
```

```python
def run_simulation(
    start_date: str = "2024-01-01",
    n_days: int = 365,
    drain: Optional[DrainConfig] = None,
    timestep_minutes: int = SIM_CONFIG["timestep_minutes"],
    sensor_interval_minutes: int = SIM_CONFIG["sensor_interval_minutes"],
    seed: Optional[int] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Runs the full pipeline for a single drain: Rainfall -> Environment ->
    Demand -> Hydraulics -> down-sampled sensor readings.

    Returns (sensor_readings_df, rain_event_log_df).
    `sensor_readings_df` is resampled to `sensor_interval_minutes` to mimic
    the real radar node's deep-sleep duty cycle.
    """
    rng = random.Random(seed)
    np_seed_env = None if seed is None else seed + 1
    np_seed_drain = None if seed is None else seed + 2

    timeline = pd.date_range(
        start=start_date,
        periods=int(round(n_days * 24 * 60 / timestep_minutes)),
        freq=f"{timestep_minutes}min",
    )

    # 1. Rainfall
    rain_engine = RainfallEngine(timestep_minutes)
    rain_mm, event_log = rain_engine.simulate(start_date, n_days, rng)
    rain_intensity_mmph = rain_mm / (timestep_minutes / 60.0)

    # 2. Environment
    env_engine = EnvironmentEngine(np.random.default_rng(np_seed_env))
    env_df = env_engine.simulate(timeline, rain_intensity_mmph)

    # 3. Demand
    hour_of_day = timeline.hour.values + timeline.minute.values / 60.0
    demand_idx = DemandEngine.demand_index(hour_of_day, env_df["E"].values)

    # 4. Hydraulics (single drain)
    if drain is None:
        drain = DrainConfig()
    drain_sim = DrainSimulator(drain, timestep_minutes, rng_seed=np_seed_drain)
    hydro_df = drain_sim.simulate(timeline, rain_intensity_mmph, demand_idx)

    full_df = env_df.copy()
    full_df["rain_mm"] = rain_mm
    full_df["rain_intensity_mmph"] = rain_intensity_mmph
    full_df["demand_index"] = demand_idx
    full_df = pd.concat([full_df, hydro_df], axis=1)
    full_df.drop(columns=["E"], inplace=True)
    full_df.index.name = "timestamp"

    # Down-sample to the sensor's actual reporting interval (deep-sleep duty cycle)
    if sensor_interval_minutes > timestep_minutes:
        step_ratio = sensor_interval_minutes // timestep_minutes
        sensor_df = full_df.iloc[::step_ratio].copy()
    else:
        sensor_df = full_df

    events_df = pd.DataFrame(event_log)
    return sensor_df, events_df
```
